```python
members = {"M101", "M102", "M103"}   # Valid members

books = {
    "B1": {"title": "Python Basics", "available": True, "days_left": 0},
    "B2": {"title": "Data Science 101", "available": True, "days_left": 0},
    "B3": {"title": "Machine Learning", "available": True, "days_left": 0}
}

LENDING_PERIOD = 7          # Book must be returned in 7 days
LATE_FEE_PER_DAY = 3        # $3 per late day

# MENU
print("==============================")
print("        LIBRARY MENU")
print("==============================")
print("1. Issue Book")
print("2. Return Book")
print("3. Search Book")
print("4. Exit")
print("==============================")

# MAIN LOOP
while True:
    choice = input("\nEnter your choice: ")


    # ISSUE BOOK
    if choice == "1":
        print("\n--- ISSUE BOOK ---")
        member_id = input("Enter Member ID: ").upper()
        book_id = input("Enter Book ID: ").upper()

        if member_id not in members:
            print("Invalid Member ID")
        elif book_id not in books:
            print("Book ID not found")
        elif books[book_id]["available"] == False:
            print("Book already issued")
        else:
            books[book_id]["available"] = False
            books[book_id]["days_left"] = LENDING_PERIOD

            print(f"Book '{books[book_id]['title']}' issued")
            print("Due in:", LENDING_PERIOD, "days")

    # RETURN BOOK
    elif choice == "2":
        print("\n--- RETURN BOOK ---")
        book_id = input("Enter Book ID: ")
        days_passed = int(input("How many days passed since issue? "))

        if book_id not in books:
            print("Book ID not found")
        elif books[book_id]["available"] == True:
            print("Book was not issued")
        else:
            due_days = books[book_id]["days_left"]
            late_days = max(0, days_passed - due_days)
            late_fee = late_days * LATE_FEE_PER_DAY

            books[book_id]["available"] = True
            books[book_id]["days_left"] = 0

            print(f"Book '{books[book_id]['title']}' returned")

            if late_days > 0:
                print(f"Late by {late_days} days")
                print(f"Late Fee: ${late_fee}")
            else:
                print("Returned on time — No fee")

    # SEARCH BOOK
    elif choice == "3":
        print("\n--- SEARCH BOOK ---")
        book_id = input("Enter Book ID: ")

        if book_id not in books:
            print(" Book ID not found")
        else:
            book = books[book_id]
            status = "Available" if book["available"] else "Issued"

            print("Title:", book["title"])
            print("Status:", status)

            if status == "Issued":
                print("Due in:", book["days_left"], "days")

    # EXIT
    elif choice == "4":
        print("Exiting system...")
        break

    else:
        print("Invalid choice. Try again.")
```
