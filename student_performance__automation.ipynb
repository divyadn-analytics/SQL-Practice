```python
import numpy as np          # For numeric calculations
import pandas as pd         # For DataFrame table

# Raw student data with some missing scores
student_records = {
    "Andy":  [85, 90, None, 88],
    "Bob":    [70, None, 75, 80],
    "Hari":[None, 95, 92, 90],
    "Mat":  [88, None, None, 91],
    "Pete":  [100, 98, 97, None]
}

Missing_value = 0           # Replace None with 0

student_summary = []        # Store processed student results

# Process each student's scores
for name, scores in student_records.items():
    cleaned = []            # Cleaned score list

    for s in scores:        # Replace None with Misssing_value
        cleaned.append(Missing_value if s is None else s)

    arr = np.array(cleaned, dtype=float)   # Convert to NumPy array
    mean_score = np.mean(arr)              # Calculate mean
    max_score = np.max(arr)                # Calculate max

    # Assign letter grade
    if mean_score >= 90:
        grade = "A"
    elif mean_score >= 80:
        grade = "B"
    elif mean_score >= 70:
        grade = "C"
    elif mean_score >= 60:
        grade = "D"
    else:
        grade = "F"

    # Add student summary to list
    student_summary.append({
        "Name": name,
        "Scores": cleaned,
        "Mean": mean_score,
        "Max": max_score,
        "Grade": grade
    })

# Convert results into a DataFrame
df = pd.DataFrame(student_summary)

class_average = df["Mean"].mean()          # Class average
top_student_row = df.loc[df["Mean"].idxmax()]  # Top student row
top_student_name = top_student_row["Name"]     # Top student name
top_student_mean = top_student_row["Mean"]     # Top student mean

running = True   # Menu loop flag

# Interactive menu
while running:
    print("===== Student Performance Tracker Menu =====")
    print("1. View Global Class Report")
    print("2. Lookup Individual Student by Name")
    print("3. Exit")

    choice = input("Enter your choice (1/2/3): ").strip()

    if choice == "1":   # Show full report
        print("\n========== GLOBAL CLASS REPORT ==========")
        print(df.to_string(index=False))
        print("\nClass Average (Mean of Means):", round(class_average, 2))
        print("Top Student:", top_student_name, "with mean score", round(top_student_mean, 2))
        print("=========================================\n")

    elif choice == "2":   # Search student
        name = input("\nEnter student name to lookup: ").strip()
        result = df[df["Name"].str.lower() == name.lower()]
        if result.empty:
            print("No record found for student:", name)
        else:
            print("\n------ Student Record ------")
            print(result.to_string(index=False))
            print("----------------------------\n")

    elif choice == "3":   # Exit program
        print("\nExiting Student Performance Tracker. Goodbye!\n")
        running = False

    else:
        print("Invalid choice. Please enter 1, 2, or 3.\n")
```
