```python

zones = {"560001", "560002", "560003", "560004"}

# List of delivery partners / drivers
drivers = ["Alex Rivera", "Jordan Smith", "Taylor Chen"]

# Inventory with clear statuses ("Sorted" / "In Warehouse")
inventory = {
    "package1": {"client": "Amazon", "destination": "560001", "weight": "5", "status": "In Warehouse", "driver": "Unassigned"},
    "package2": {"client": "Nike", "destination": "560003", "weight": "3", "status": "In Warehouse", "driver": "Unassigned"},
    "package3": {"client": "Apple", "destination": "560004", "weight": "8", "status": "In Warehouse", "driver": "Unassigned"}
}

# Delivery Van Loading Queue
van_queue = []

# Automatic package ID counter
counter = 4

#  Main continuous loop
while True:
    print("\n--- MENU ---")
    print("1. Add New Package")
    print("2. View All Packages")
    print("3. Load Package into Van")
    print("4. Deliver Next Package")
    print("5. View Drivers List")
    print("6. Exit")

    #  Error handling on input choices
    choice = input("\nEnter choice (1-6): ").strip()

  # add new parcel
    if choice == "1":
        print("\n[+] ADD NEW PACKAGE")
        client = input("Enter Client Name: ").strip()
        postal_code = input("Enter Destination Postal Code: ").strip()

        #  Postal code validation
        if postal_code not in zones:
            print(" Invalid Zone! We do not deliver to postal code " + postal_code)
            continue

        weight = input("Enter Package Weight (kg): ").strip()

        # Auto ID generation
        pkg_id = "package" + str(counter)
        counter = counter + 1

        # Initial status set to "Label Generated"
        inventory[pkg_id] = {
            "client": client,
            "destination": postal_code,
            "weight": weight,
            "status": "Label Generated",
            "driver": "Unassigned"
        }
        print(" Success! Registered package with ID: " + pkg_id + " (Status: Label Generated)")

   #inventory mgmt  #(An inventory audit is an official check of all items in a warehouse to make sure digital records match physical stock.
   #In our code, Option 2 acts as an automated audit by printing out a
   #complete list of every package in our system—showing its ID, destination, status, and driver so nothing gets lost.")
    elif choice == "2":
        print("\n--- INVENTORY AUDIT ---")
        for pkg_id in inventory:
            info = inventory[pkg_id]
            client = info["client"]
            zone = info["destination"]
            status = info["status"]
            driver = info["driver"]

            print("ID: " + pkg_id + " | Client: " + client + " | Zone: " + zone + " | Status: " + status + " | Driver: " + driver)


    # MOVE FROM INVENTORY TO VAN QUEUE ("In Warehouse" -> "Out for Delivery")

    elif choice == "3":
        print("\n[>] LOAD PACKAGE INTO VAN")
        selected_id = input("Enter Package ID to load (e.g., package1): ").strip().lower()

        if selected_id in inventory:
            # Assign a driver from our list
            assigned_driver = drivers[0]

            #  Change status to "Out for Delivery"
            inventory[selected_id]["status"] = "Out for Delivery"
            inventory[selected_id]["driver"] = assigned_driver

            # Add to van sequence
            van_queue.append(selected_id)

            print(" Package " + selected_id + " loaded into van for driver " + assigned_driver + "!")
        else:
            print(" Error: Package ID not found!")


    # 4. PROCESS LIVE DELIVERY (Using pop())

    elif choice == "4":
        print("\n[] DELIVER PACKAGE")

        #  Error handling for empty van
        if len(van_queue) == 0:
            print(" Van is empty! No packages ready for delivery.")
        else:
            # Pop package from queue
            delivered_id = van_queue.pop(0)

            # Update status to "Delivered"
            inventory[delivered_id]["status"] = "Delivered"

            pkg_info = inventory[delivered_id]
            client_name = pkg_info["client"]

            print(" Package " + delivered_id + " delivered to " + client_name + "! Status: Delivered")


    #  VIEW DRIVERS

    elif choice == "5":
        print("\n--- ACTIVE DELIVERY DRIVERS ---")
        for driver in drivers:
            print(" - " + driver)


    # EXIT OPTION

    elif choice == "6":
        print("\nExiting program. Goodbye!")
        #Correct use of break
        break

   # Handle unexpected menu inputs
    else:
        print(" Invalid menu choice. Please select between 1 and 6.")
```
