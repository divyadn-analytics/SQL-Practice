```python

import numpy as np
import pandas as pd

students = {
    "Ram":[85,90,None],
    "Karthik":[70,None,75],
    "Hari":[95,92,98],
    "Shiv":[60,65,None]
}

student_results = []

for name, scores in students.items():

    # Remove None values
    valid_scores = [s for s in scores if pd.notna(s)]

    # Calculate mean and median
    mean_value = np.mean(valid_scores)
    median_value = np.median(valid_scores)

    # Store results
    student_result = [name, mean_value, median_value]
    student_results.append(student_result)

df = pd.DataFrame(student_results, columns=["Name", "Mean", "Median"])
print(df)

# Mode is not calculated for this dataset because none of the students valid scores include
#repeated values. therefore, there is no mode.
```
