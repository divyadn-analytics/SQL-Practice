```python
zones={"560001","560002","560003","560004"}
package1 = { "client": "Amazon","destination": "560001","weight": 5,"status": "Sorted" }
package2 = { "client": "Nike", "destination": "560003", "weight": 3, "status": "Sorted" }
package3 = { "client": "Apple", "destination": "560004", "weight": 8, "status": "Sorted"}

packages = [package1, package2, package3]

print("Checking zones...")

for pkg in packages:
    if pkg["destination"] in zones:
        print(pkg["client"], "→ valid zone")
    else:
        print(pkg["client"], "→ invalid zone")

loading_order = []

for pkg in packages:
    loading_order.append(pkg)

print("Loading Order")

count = 1

for pkg in loading_order:
    print(count, pkg["client"], pkg["destination"])
    count += 1

client = input("Enter client name: ")
destination = input("Enter destination postal code: ")
weight_text = input("Enter weight (kg): ")

try:
    weight = float(weight_text)
except:
    print("Invalid weight.")
    weight = None

if destination not in zones:
    print("Postal code not supported.")
else:
    new_package = { "client": client, "destination": destination, "weight": weight, "status": "Sorted" }
    packages.append(new_package)
    loading_order.append(new_package)
    print("Package added.")
print("Packages:")

for i in range(len(packages)):
    pkg = packages[i]
    print(i, pkg["client"], pkg["destination"], pkg["status"])

index_text = input("Enter package number to update: ")

try:
    index = int(index_text)
except:
    print("Invalid number.")
    index = None

statuses = {"Sorted", "Dispatched", "Delivered"}

print("Valid statuses:", statuses)
new_status = input("Enter new status: ")

if new_status not in statuses:
    print("Invalid status.")
else:
    packages[index]["status"] = new_status
    print("Status updated!")
    print("Current packages:")

for pkg in packages:
    print(pkg["client"], pkg["destination"], pkg["status"])

remaining_packages = []

for pkg in packages:
    if pkg["status"] != "Delivered":
        remaining_packages.append(pkg)

packages = remaining_packages

print("Delivered packages removed.")
```
